```python
import pandas as pd
```

```python
morrowplots_df = pd.read_csv("../data/raw/morrowplots.csv")
```

```python
morrowplots_df
```

```python
mpcorn_df = morrowplots_df[morrowplots_df['corn']==True]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 6))
# seaborn makes it easier to handle grouped aesthetics like color + linetype
sns.lineplot(
    data=mpcorn_df,
    x="year",
    y="yield_bush",
    hue="rotation",
    style="treated",
    lw=2
)
plt.title("Morrow Plots Corn Yield Trends in Bushels per Acre")
plt.xlabel("Year")
plt.ylabel("Yield (bushels per acre)")
plt.legend(title="Rotation / Treated", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
